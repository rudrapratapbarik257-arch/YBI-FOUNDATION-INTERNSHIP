# Python Dictionary

A dictionary stores data as key-value pairs. Each key must be unique.

## 1. Creating a Dictionary


```python
student = {"name": "Rahul", "age": 20, "course": "BCA"}
print(student)
```

## 2. Accessing Values


```python
student = {"name": "Rahul", "age": 20, "course": "BCA"}
print(student["name"])
print(student["age"])
```

## 3. Using get()


```python
student = {"name": "Rahul", "age": 20}
print(student.get("name"))
print(student.get("city", "Not found"))
```

## 4. Adding Items


```python
student = {"name": "Rahul", "age": 20}
student["city"] = "Bhubaneswar"
print(student)
```

## 5. Updating Values


```python
student = {"name": "Rahul", "age": 20}
student["age"] = 21
print(student)
```

## 6. update() Method


```python
student = {"name": "Rahul", "age": 20}
student.update({"course": "BCA", "age": 21})
print(student)
```

## 7. pop() Method


```python
student = {"name": "Rahul", "age": 20, "course": "BCA"}
removed = student.pop("age")
print("Removed value:", removed)
print(student)
```

## 8. popitem() Method


```python
student = {"name": "Rahul", "age": 20}
removed = student.popitem()
print("Removed item:", removed)
print(student)
```

## 9. del Statement


```python
student = {"name": "Rahul", "age": 20, "course": "BCA"}
del student["age"]
print(student)
```

## 10. clear() Method


```python
student = {"name": "Rahul", "age": 20}
student.clear()
print(student)
```

## 11. keys() Method


```python
student = {"name": "Rahul", "age": 20, "course": "BCA"}
print(student.keys())
```

## 12. values() Method


```python
student = {"name": "Rahul", "age": 20, "course": "BCA"}
print(student.values())
```

## 13. items() Method


```python
student = {"name": "Rahul", "age": 20, "course": "BCA"}
print(student.items())
```

## 14. Loop Through Dictionary


```python
student = {"name": "Rahul", "age": 20, "course": "BCA"}
for key in student:
    print(key, ":", student[key])
```

## 15. Loop Through keys and values


```python
student = {"name": "Rahul", "age": 20, "course": "BCA"}
for key, value in student.items():
    print(key, ":", value)
```

## 16. Check if a Key Exists


```python
student = {"name": "Rahul", "age": 20}
if "name" in student:
    print("Name key exists")
else:
    print("Name key does not exist")
```

## 17. len() Function


```python
student = {"name": "Rahul", "age": 20, "course": "BCA"}
print("Number of items:", len(student))
```

## 18. copy() Method


```python
original = {"a": 1, "b": 2}
new_dict = original.copy()
print("Original:", original)
print("Copy:", new_dict)
```

## 19. Nested Dictionary


```python
students = {
    "student1": {"name": "Rahul", "age": 20},
    "student2": {"name": "Aman", "age": 21}
}
print(students["student1"]["name"])
print(students["student2"]["age"])
```

## 20. Dictionary Comprehension


```python
squares = {x: x * x for x in range(1, 6)}
print(squares)
```

## 21. setdefault() Method


```python
student = {"name": "Rahul"}
student.setdefault("course", "BCA")
print(student)
```

## 22. fromkeys() Method


```python
keys = ["name", "age", "course"]
student = dict.fromkeys(keys, "Unknown")
print(student)
```

## 23. Merge Two Dictionaries


```python
first = {"a": 1, "b": 2}
second = {"b": 3, "c": 4}
merged = first | second
print(merged)
```

## 24. Practical Student Record


```python
student = {"name": "Subham", "course": "BCA", "semester": 3, "marks": 85}
for key, value in student.items():
    print(f"{key}: {value}")
```
