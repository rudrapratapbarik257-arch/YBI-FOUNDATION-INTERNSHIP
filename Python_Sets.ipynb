# Python Sets

Examples of Python set creation, methods, and set operations.

## Create a set

```python
numbers = {1, 2, 3, 4}
print(numbers)
```

## Set removes duplicates

```python
numbers = {1, 2, 2, 3, 3, 3}
print(numbers)
```

## Create set using set()

```python
items = set([1, 2, 3])
print(items)
```

## Empty set

```python
empty_set = set()
print(empty_set)
```

## Add an element

```python
fruits = {'apple', 'banana'}
fruits.add('mango')
print(fruits)
```

## Add multiple elements

```python
numbers = {1, 2}
numbers.update([3, 4, 5])
print(numbers)
```

## Remove an element

```python
fruits = {'apple', 'banana', 'mango'}
fruits.remove('banana')
print(fruits)
```

## Discard an element

```python
fruits = {'apple', 'banana'}
fruits.discard('orange')
print(fruits)
```

## Pop an element

```python
numbers = {10, 20, 30}
removed = numbers.pop()
print('Removed:', removed)
print('Remaining:', numbers)
```

## Clear a set

```python
numbers = {1, 2, 3}
numbers.clear()
print(numbers)
```

## Check membership

```python
colors = {'red', 'green', 'blue'}
print('red' in colors)
print('yellow' in colors)
```

## Set length

```python
numbers = {10, 20, 30, 40}
print(len(numbers))
```

## Union

```python
a = {1, 2, 3}
b = {3, 4, 5}
print(a.union(b))
```

## Intersection

```python
a = {1, 2, 3}
b = {2, 3, 4}
print(a.intersection(b))
```

## Difference

```python
a = {1, 2, 3}
b = {2, 3, 4}
print(a.difference(b))
```

## Symmetric difference

```python
a = {1, 2, 3}
b = {3, 4, 5}
print(a.symmetric_difference(b))
```

## Subset

```python
a = {1, 2}
b = {1, 2, 3, 4}
print(a.issubset(b))
```

## Superset

```python
a = {1, 2, 3, 4}
b = {1, 2}
print(a.issuperset(b))
```

## Disjoint sets

```python
a = {1, 2}
b = {3, 4}
print(a.isdisjoint(b))
```

## Loop through a set

```python
fruits = {'apple', 'banana', 'mango'}
for fruit in fruits:
    print(fruit)
```

## Set comprehension

```python
squares = {x * x for x in range(1, 6)}
print(squares)
```

## Convert list to set

```python
values = [1, 2, 2, 3, 4, 4]
unique_values = set(values)
print(unique_values)
```
