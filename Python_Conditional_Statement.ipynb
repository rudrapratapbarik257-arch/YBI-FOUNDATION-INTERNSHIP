# Python Conditional Statements

Examples of Python conditional statements.

## Simple if


```python
age = 20
if age >= 18:
    print("You are eligible to vote.")
```

## if-else


```python
number = 7
if number % 2 == 0:
    print("Even number")
else:
    print("Odd number")
```

## if-elif-else


```python
marks = 75
if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
```

## Positive, Negative, or Zero


```python
number = -5
if number > 0:
    print("Positive number")
elif number < 0:
    print("Negative number")
else:
    print("Zero")
```

## Largest of Two Numbers


```python
a, b = 25, 40
if a > b:
    print("a is larger")
elif b > a:
    print("b is larger")
else:
    print("Both are equal")
```

## Largest of Three Numbers


```python
a, b, c = 12, 35, 20
if a >= b and a >= c:
    print("Largest:", a)
elif b >= a and b >= c:
    print("Largest:", b)
else:
    print("Largest:", c)
```

## Nested if


```python
age = 22
has_id = True
if age >= 18:
    if has_id:
        print("Entry allowed")
    else:
        print("Please show your ID")
else:
    print("Entry not allowed")
```

## Voting Eligibility


```python
age = 19
if age >= 18:
    print("Eligible to vote")
else:
    print("Not eligible to vote")
```

## Leap Year


```python
year = 2024
if year % 400 == 0 or (year % 4 == 0 and year % 100 != 0):
    print("Leap year")
else:
    print("Not a leap year")
```

## Divisibility by 3 and 5


```python
number = 30
if number % 3 == 0 and number % 5 == 0:
    print("Divisible by both 3 and 5")
elif number % 3 == 0:
    print("Divisible by 3")
elif number % 5 == 0:
    print("Divisible by 5")
else:
    print("Not divisible by 3 or 5")
```

## Logical Operators


```python
age = 21
has_ticket = True
if age >= 18 and has_ticket:
    print("You can enter")
else:
    print("You cannot enter")
```

## Membership Operator


```python
language = "Python"
if "Py" in language:
    print("Text found")
else:
    print("Text not found")
```

## Short-hand if


```python
number = 10
if number > 0: print("Positive number")
```

## Conditional Expression


```python
number = 9
result = "Even" if number % 2 == 0 else "Odd"
print(result)
```

## Character Type


```python
character = "A"
if character.isalpha():
    print("Alphabet")
elif character.isdigit():
    print("Digit")
else:
    print("Special character")
```

## Simple Calculator


```python
a, b = 12, 4
operator = "*"
if operator == "+":
    print(a + b)
elif operator == "-":
    print(a - b)
elif operator == "*":
    print(a * b)
elif operator == "/":
    if b != 0:
        print(a / b)
    else:
        print("Cannot divide by zero")
else:
    print("Invalid operator")
```

## pass Statement


```python
number = 5
if number > 0:
    pass
else:
    print("Number is not positive")
```
