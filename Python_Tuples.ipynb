# Python Tuples

Examples of creating, accessing, unpacking, and using tuples in Python.

## Creating Tuples


```python
fruits = ("apple", "banana", "cherry")
print(fruits)
```

## Tuple with Different Data Types


```python
person = ("Aman", 21, 5.8, True)
print(person)
```

## Accessing Tuple Items


```python
colors = ("red", "green", "blue")
print(colors[0])
print(colors[2])
```

## Negative Indexing


```python
numbers = (10, 20, 30, 40)
print(numbers[-1])
print(numbers[-2])
```

## Tuple Slicing


```python
numbers = (0, 1, 2, 3, 4, 5)
print(numbers[1:4])
print(numbers[:3])
print(numbers[3:])
```

## Tuple Length


```python
items = ("pen", "book", "pencil")
print(len(items))
```

## Tuple with One Item


```python
single = ("apple",)
print(type(single))
```

## Tuples Are Immutable


```python
fruits = ("apple", "banana", "cherry")
# Tuples cannot be changed after creation.
print(fruits)
```

## Loop Through a Tuple


```python
fruits = ("apple", "banana", "cherry")
for fruit in fruits:
    print(fruit)
```

## Check if an Item Exists


```python
fruits = ("apple", "banana", "cherry")
if "banana" in fruits:
    print("Banana is present")
else:
    print("Banana is not present")
```

## Tuple Concatenation


```python
a = (1, 2, 3)
b = (4, 5)
combined = a + b
print(combined)
```

## Repeating a Tuple


```python
numbers = (1, 2)
print(numbers * 3)
```

## Tuple Methods: count and index


```python
numbers = (1, 2, 2, 3, 2)
print(numbers.count(2))
print(numbers.index(3))
```

## Tuple Unpacking


```python
person = ("Riya", 20, "India")
name, age, country = person
print(name)
print(age)
print(country)
```

## Extended Unpacking


```python
numbers = (1, 2, 3, 4, 5)
first, *middle, last = numbers
print(first)
print(middle)
print(last)
```

## Convert List to Tuple


```python
colors_list = ["red", "green", "blue"]
colors_tuple = tuple(colors_list)
print(colors_tuple)
```

## Convert Tuple to List


```python
numbers_tuple = (1, 2, 3)
numbers_list = list(numbers_tuple)
numbers_list.append(4)
print(numbers_list)
```

## Nested Tuples


```python
matrix = ((1, 2), (3, 4), (5, 6))
print(matrix[0])
print(matrix[1][1])
```

## Find Minimum and Maximum


```python
numbers = (8, 3, 12, 5, 1)
print("Minimum:", min(numbers))
print("Maximum:", max(numbers))
```

## Sort Values Using sorted()


```python
numbers = (5, 2, 9, 1, 7)
print(sorted(numbers))
```

## Tuple Membership


```python
values = ("Python", "Java", "C++")
print("Python" in values)
print("Ruby" not in values)
```

## Practice Example


```python
scores = (75, 88, 92, 67, 88)
print("Total scores:", len(scores))
print("Highest score:", max(scores))
print("Number of 88s:", scores.count(88))
```
