# Python Loops

Beginner-friendly examples of `for` loops, `while` loops, and loop control statements.

## 1. for loop


```python
for i in range(1, 6):
    print(i)
```

## 2. for loop with a list


```python
fruits = ["Apple", "Banana", "Mango"]
for fruit in fruits:
    print(fruit)
```

## 3. range with start and stop


```python
for i in range(2, 11, 2):
    print(i)
```

## 4. while loop


```python
i = 1
while i <= 5:
    print(i)
    i += 1
```

## 5. Sum of first 10 numbers


```python
total = 0
for i in range(1, 11):
    total += i
print("Sum:", total)
```

## 6. Multiplication table


```python
number = 5
for i in range(1, 11):
    print(number, "x", i, "=", number * i)
```

## 7. break statement


```python
for i in range(1, 10):
    if i == 5:
        break
    print(i)
```

## 8. continue statement


```python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

## 9. pass statement


```python
for i in range(3):
    pass
print("Loop finished")
```

## 10. Nested loops


```python
for i in range(1, 4):
    for j in range(1, 4):
        print(i, j)
```

## 11. Loop through a string


```python
word = "Python"
for letter in word:
    print(letter)
```

## 12. Even numbers


```python
for i in range(2, 21):
    if i % 2 == 0:
        print(i)
```

## 13. Odd numbers


```python
for i in range(1, 20, 2):
    print(i)
```

## 14. Factorial using a loop


```python
number = 5
factorial = 1
for i in range(1, number + 1):
    factorial *= i
print("Factorial:", factorial)
```

## 15. Reverse counting with while


```python
i = 5
while i >= 1:
    print(i)
    i -= 1
```

## 16. else with for loop


```python
for i in range(1, 4):
    print(i)
else:
    print("Loop completed")
```

## 17. Find a number in a list


```python
numbers = [10, 20, 30, 40]
target = 30
for number in numbers:
    if number == target:
        print("Found", target)
        break
else:
    print("Not found")
```

## 18. Star pattern


```python
for i in range(1, 6):
    print("*" * i)
```
