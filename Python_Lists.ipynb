# Python Lists

Examples of the main Python List concepts and methods.

## 1. Creating a List


```python
fruits = ["Apple", "Banana", "Mango", "Orange"]
print(fruits)
```

## 2. Different Data Types


```python
my_list = [10, "Python", 3.14, True]
print(my_list)
```

## 3. Accessing Elements


```python
fruits = ["Apple", "Banana", "Mango", "Orange"]
print(fruits[0])
print(fruits[2])
```

## 4. Negative Indexing


```python
fruits = ["Apple", "Banana", "Mango", "Orange"]
print(fruits[-1])
print(fruits[-2])
```

## 5. List Slicing


```python
numbers = [10, 20, 30, 40, 50]
print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])
```

## 6. Changing Elements


```python
fruits = ["Apple", "Banana", "Mango"]
fruits[1] = "Grapes"
print(fruits)
```

## 7. append()


```python
fruits = ["Apple", "Banana"]
fruits.append("Mango")
print(fruits)
```

## 8. insert()


```python
fruits = ["Apple", "Mango"]
fruits.insert(1, "Banana")
print(fruits)
```

## 9. extend()


```python
fruits = ["Apple", "Banana"]
more_fruits = ["Mango", "Orange"]
fruits.extend(more_fruits)
print(fruits)
```

## 10. remove()


```python
fruits = ["Apple", "Banana", "Mango"]
fruits.remove("Banana")
print(fruits)
```

## 11. pop()


```python
fruits = ["Apple", "Banana", "Mango"]
removed = fruits.pop(1)
print("Removed:", removed)
print(fruits)
```

## 12. del


```python
numbers = [10, 20, 30, 40]
del numbers[1]
print(numbers)
```

## 13. clear()


```python
numbers = [10, 20, 30, 40]
numbers.clear()
print(numbers)
```

## 14. len()


```python
fruits = ["Apple", "Banana", "Mango", "Orange"]
print("Length:", len(fruits))
```

## 15. count()


```python
numbers = [10, 20, 10, 30, 10, 40]
print(numbers.count(10))
```

## 16. index()


```python
fruits = ["Apple", "Banana", "Mango", "Orange"]
print(fruits.index("Mango"))
```

## 17. sort()


```python
numbers = [50, 10, 40, 20, 30]
numbers.sort()
print(numbers)
```

## 18. reverse()


```python
numbers = [10, 20, 30, 40, 50]
numbers.reverse()
print(numbers)
```

## 19. Membership Operators


```python
fruits = ["Apple", "Banana", "Mango"]
print("Apple" in fruits)
print("Orange" not in fruits)
```

## 20. copy()


```python
original = [10, 20, 30]
copy_list = original.copy()
print("Original:", original)
print("Copy:", copy_list)
```

## 21. Loop Through a List


```python
fruits = ["Apple", "Banana", "Mango"]
for fruit in fruits:
    print(fruit)
```

## 22. List Comprehension


```python
numbers = [1, 2, 3, 4, 5]
squares = [x * x for x in numbers]
print(squares)
```

## 23. List Comprehension with Condition


```python
numbers = [1, 2, 3, 4, 5, 6]
even_numbers = [x for x in numbers if x % 2 == 0]
print(even_numbers)
```

## 24. Nested List


```python
students = [
    ["Rahul", 20],
    ["Aman", 21],
    ["Ravi", 19]
]
print(students)
print(students[0][0])
```

## 25. Maximum and Minimum


```python
numbers = [10, 25, 5, 40, 15]
print("Maximum:", max(numbers))
print("Minimum:", min(numbers))
```

## 26. Sum of List Elements


```python
numbers = [10, 20, 30, 40, 50]
print("Sum:", sum(numbers))
```
